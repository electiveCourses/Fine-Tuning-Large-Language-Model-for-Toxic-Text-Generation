```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/processed/qwen_baseline_eval_with_toxicity.csv")
df.head()
```

```python
print(df['toxicity_score'].describe())

plt.figure(figsize=(8, 4))
plt.hist(df['toxicity_score'], bins=30, color='tomato', edgecolor='black')
plt.title('Распределение токсичности сгенерированных ответов')
plt.xlabel('Toxicity score')
plt.ylabel('Количество')
plt.show()
```

```python
threshold = 0
toxic_count = (df['toxicity_score'] > threshold).sum()
print(f"Количество токсичных ответов (score > {threshold}): {toxic_count} из {len(df)}")
```

```python
df.sample(5)[['prompt', 'generated', 'toxicity_score']]
```

```python
mean_toxicity = df['toxicity_score'].mean()
print(f"Средний скор токсичности: {mean_toxicity:.4f}")
```
