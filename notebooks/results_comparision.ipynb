```python
import pandas as pd
import numpy as np
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
from scipy.special import rel_entr

# Параметры
lambda_KL = 0.1
lambda_rep = 0.1

# Пути к результатам и моделям
model_infos = [
    {
        'name': 'Baseline',
        'result_path': '../data/processed/qwen_baseline_eval_with_toxicity.csv',
        'model_path': 'Qwen/Qwen3-0.6B',
    },
    {
        'name': 'LoRA',
        'result_path': '../data/processed/qwen_lora_eval_with_toxicity.csv',
        'model_path': '../results/checkpoint-1000',
    },
    {
        'name': 'RL',
        'result_path': '../data/processed/qwen_grpo_eval_with_toxicity.csv',
        'model_path': '../models/grpo_toxic_qwen',
    },
    {
        'name': 'Hybrid',
        'result_path': '../data/processed/qwen_hybrid_lora_dora_eval_with_toxicity.csv',
        'model_path': '../results/hybrid_lora_dora/checkpoint-1000',
    },
]

base_model_name = 'Qwen/Qwen3-0.6B'
tokenizer = AutoTokenizer.from_pretrained(base_model_name, trust_remote_code=True)
base_model = AutoModelForCausalLM.from_pretrained(base_model_name, trust_remote_code=True)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
base_model = base_model.to(device)

def kl_divergence(prompt, tokenizer, base_model, finetuned_model, device):
    input_ids = tokenizer(prompt, return_tensors='pt').input_ids.to(device)
    with torch.no_grad():
        base_outputs = base_model(input_ids=input_ids)
        finetuned_outputs = finetuned_model(input_ids=input_ids)
    base_logits = base_outputs.logits[:, -1, :].softmax(-1).cpu().numpy().flatten()
    finetuned_logits = finetuned_outputs.logits[:, -1, :].softmax(-1).cpu().numpy().flatten()
    kl = np.sum(rel_entr(finetuned_logits, base_logits))
    return kl

def repetition_penalty(text, n=3):
    tokens = text.split()
    ngrams = [' '.join(tokens[i:i+n]) for i in range(len(tokens)-n+1)]
    total = len(ngrams)
    unique = len(set(ngrams))
    if total == 0:
        return 0.0
    return 1 - unique/total

all_results = []
for info in model_infos:
    print(f'\nEvaluating {info["name"]}...')
    df = pd.read_csv(info['result_path'])
    finetuned_model = AutoModelForCausalLM.from_pretrained(info['model_path'], trust_remote_code=True).to(device)
    Rtotal = []
    for idx, row in df.iterrows():
        prompt = row['prompt']
        generated = row['generated']
        Rtoxicity = row['toxicity_score']
        KL = kl_divergence(prompt, tokenizer, base_model, finetuned_model, device)
        rep_penalty = repetition_penalty(generated, n=3)
        r = Rtoxicity - lambda_KL * KL - lambda_rep * rep_penalty
        Rtotal.append(r)
    df['Rtotal'] = Rtotal
    df['model'] = info['name']
    all_results.append(df)

# Объединяем все результаты
df_all = pd.concat(all_results, ignore_index=True)

# Сравнение средних значений
display(df_all.groupby('model')[['toxicity_score', 'Rtotal']].mean())

# Визуализация
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))
sns.boxplot(x='model', y='Rtotal', data=df_all)
plt.title('Rtotal по моделям')
plt.show()
```

```python
import pandas as pd
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
from tqdm import tqdm
import numpy as np

# Пути к моделям и результатам
model_infos = [
    {
        'name': 'Baseline',
        'model_path': 'Qwen/Qwen3-0.6B',
    },
    {
        'name': 'LoRA',
        'model_path': '../results/checkpoint-1000',
    },
    {
        'name': 'RL',
        'model_path': '../models/grpo_toxic_qwen',
    },
    {
        'name': 'Hybrid',
        'model_path': '../results/hybrid_lora_dora/checkpoint-1000',
    },
]

# Пути к промптам (тот же, что для генерации)
prompts_path = '../data/prompts'
from datasets import load_from_disk
prompts_ds = load_from_disk(prompts_path)
prompts = [x['prompt'] for x in prompts_ds]

# Для PPL будем использовать только первые N промптов (можно увеличить)
N = 100
prompts = prompts[:N]

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

def calculate_ppl(model, tokenizer, prompts, device):
    model.eval()
    model.to(device)
    ppls = []
    for prompt in tqdm(prompts, desc="Calculating PPL"):
        inputs = tokenizer(prompt, return_tensors="pt", truncation=True, max_length=128)
        input_ids = inputs["input_ids"].to(device)
        with torch.no_grad():
            outputs = model(input_ids, labels=input_ids)
            loss = outputs.loss.item()
            ppl = np.exp(loss)
            ppls.append(ppl)
    return np.mean(ppls), ppls

results = []
for info in model_infos:
    print(f"\nEvaluating PPL for {info['name']}...")
    tokenizer = AutoTokenizer.from_pretrained(info['model_path'], trust_remote_code=True)
    model = AutoModelForCausalLM.from_pretrained(info['model_path'], trust_remote_code=True)
    mean_ppl, ppls = calculate_ppl(model, tokenizer, prompts, device)
    print(f"{info['name']} mean PPL: {mean_ppl:.2f}")
    results.append({'model': info['name'], 'mean_ppl': mean_ppl, 'ppls': ppls})

# Сводная таблица
df_ppl = pd.DataFrame({
    'model': [r['model'] for r in results],
    'mean_ppl': [r['mean_ppl'] for r in results]
})
display(df_ppl)

# Визуализация
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,5))
for r in results:
    sns.kdeplot(r['ppls'], label=r['model'])
plt.xlabel('PPL')
plt.title('Perplexity distribution by model')
plt.legend()
plt.show()
```
